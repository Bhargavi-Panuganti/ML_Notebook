```python
import numpy as np

data = np.array([12, 15, 18, 20, 22, 25, 30, 35, 40])
```

```python
print(np.mean(data))
```

```python
print(np.median(data))
```

```python
print(np.max(data))
print(np.min(data))
print(np.max(data)-np.min(data))
```

```python
## Variance - Measures how far data points are from the mean.
print(np.var(data))
```

```python
print(np.std(data))
```

```python
# DETECTING OUTLIERS

import numpy as np

# Sample dataset
data = np.array([10, 12, 15, 18, 20, 22, 25, 500])

# --------------------------------------------------
# Step 1: Calculate First Quartile (Q1)
# Q1 = 25th percentile
# 25% of the data lies below this value
# This means:
# Around 25% of the data lies below 14.25.
# --------------------------------------------------
Q1 = np.percentile(data, 25)

# --------------------------------------------------
# Step 2: Calculate Third Quartile (Q3)
# Q3 = 75th percentile
# 75% of the data lies below this value
# --------------------------------------------------
Q3 = np.percentile(data, 75)

# --------------------------------------------------
# Step 3: Calculate Interquartile Range (IQR)
# IQR represents the spread of the middle 50% of data
#
# Formula:
# IQR = Q3 - Q1
# --------------------------------------------------
IQR = Q3 - Q1

# --------------------------------------------------
# Step 4: Calculate Lower Bound
#
# Any value below this limit is considered an outlier
#
# Formula:
# Lower Bound = Q1 - 1.5 * IQR
# --------------------------------------------------
lower_bound = Q1 - 1.5 * IQR

# --------------------------------------------------
# Step 5: Calculate Upper Bound
#
# Any value above this limit is considered an outlier
#
# Formula:
# Upper Bound = Q3 + 1.5 * IQR
# --------------------------------------------------
upper_bound = Q3 + 1.5 * IQR

# --------------------------------------------------
# Step 6: Detect Outliers
#
# Select values that are:
#   less than lower_bound OR
#   greater than upper_bound
# --------------------------------------------------
outliers = data[
    (data < lower_bound) |
    (data > upper_bound)
]

# --------------------------------------------------
# Step 7: Print Results
# --------------------------------------------------
print("Dataset:", data)

print("\nQ1 (25th Percentile):", Q1)
print("Q3 (75th Percentile):", Q3)

print("\nIQR:", IQR)

print("\nLower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("\nOutliers:", outliers)
```
