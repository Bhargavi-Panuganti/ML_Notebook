```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.title("Line Chart")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
plt.plot(x, y, color='red', linewidth=3, linestyle='--', marker='*')
```

```python
categories = ['A', 'B', 'C', 'D']
values = [15, 30, 22, 40]

plt.bar(categories, values, color='steelblue')
plt.title("Bar Chart")
plt.xlabel("Category")
plt.ylabel("Value")
plt.show()
```

```python
data = np.random.randn(100)  # 100 random values from a normal distribution

plt.hist(data, bins=15, color='green', edgecolor='black')
plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))  # 1 row, 3 columns

axes[0].plot([1,2,3], [10,20,15])
axes[0].set_title("Line")

axes[1].bar(['A','B','C'], [5, 15, 10])
axes[1].set_title("Bar")

axes[2].hist(np.random.randn(100), bins=10, edgecolor='black')
axes[2].set_title("Histogram")

plt.tight_layout()  # prevents labels from overlapping
plt.show()
```

```python
x = [1, 2, 3, 4, 5]

plt.plot(x, [10, 20, 15, 30, 25], label='Sales')
plt.plot(x, [5, 15, 10, 20, 18], label='Profit')

plt.legend()  # this reads the labels and draws the box
plt.legend(loc='upper right')
plt.show()
```
