```python
data = {
    "Name":   ["Ravi", "Priya", "Arjun", "Sneha", "Kiran"],
    "Marks":  [85, 92, 78, 95, 60],
    "City":   ["Hyderabad", "Vijayawada", "Chennai", "Mumbai", "Delhi"],
    "Grade":  ["B", "A", "C", "A", "D"]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.iloc[0])
print("--------------------")
print(df.iloc[4])
print("--------------------")
print(df.iloc[-1])
```

```python
print(df.iloc[0, 0])
print(df.iloc[1, 1])
print(df.iloc[3, 2])
```

```python
df.iloc[0:3, 0:2]   # Rows 0-2, Columns 0-1
```

```python
df.iloc[:, 0:2]   # ALL rows, columns 0 and 1 only
```

```python
# Pick specific rows/cols (not a range)
df.iloc[[0, 2, 4]]          # Rows 0, 2, 4 only
df.iloc[[0, 2], [0, 3]]     # Rows 0&2, Cols 0&3
```

```python
df.loc[0]       # Row with index label 0
df.loc[3]       # Row with index label 3 → Sneha
```

```python
df.loc[2]
```

```python
print(df.loc[0, "Name"])    # → "Ravi"
print(df.loc[2, "City"])     # → "Chennai"
print(df.loc[3, "Marks"])    # → 95
```

```python
df.loc[0:3]     # Rows 0, 1, 2, 3 (3 IS included — unlike iloc!)
```

```python
df.loc[:, ["Name", "Grade"]]
```

```python
df.loc[1:3, ["Name", "Marks"]]
```

```python
# All students with Grade "A"
df.loc[df["Grade"] == "A"]
```

```python
# Show only Name and Marks of Grade A students
df.loc[df["Grade"] == "A", ["Name", "Marks"]]
```

```python
df.iloc[0:3]   # Rows 0, 1, 2 → 3 NOT included
df.loc[0:3]    # Rows 0, 1, 2, 3 → 3 IS included
```

```python
data = {
    "Name":    ["Ravi", "Priya", "Arjun", "Sneha", "Kiran", "Divya", "Rohan"],
    "Marks":   [85, 92, 78, 95, 60, 88, 72],
    "City":    ["Hyderabad", "Vijayawada", "Hyderabad", "Mumbai", "Vijayawada", "Mumbai", "Hyderabad"],
    "Grade":   ["B", "A", "C", "A", "D", "B", "C"]
}

df = pd.DataFrame(data)
```

```python
# GroupBy Syntax "" df.groupby("column")["what to calculate"].operation() ""
```

```python
df.groupby("Grade")["Marks"].mean()
```

```python
df.groupby("City")["Marks"].sum()
```

```python
df.groupby("Grade")["Name"].count()
```

```python
print(df.groupby("City")["Marks"].max())   # Top scorer per city
print("----------------------------------")
df.groupby("City")["Marks"].min()   # Lowest scorer per city
```

```python
# Multiple Operations at once
df.groupby("City")["Marks"].agg(["mean", "max", "min", "count"])
```

```python
# Group By Multiple Columns
df.groupby(["Grade", "City"])["Marks"].mean()
```

```python
import pandas as pd
import numpy as np

data = {
    "Name":   ["Ravi", "Priya", "Arjun", "Sneha", "Kiran"],
    "Marks":  [85, np.nan, 78, np.nan, 60],
    "City":   ["Hyderabad", "Vijayawada", np.nan, "Mumbai", np.nan],
    "Grade":  ["B", "A", np.nan, "A", "D"]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.isnull().sum()   # Count of NaN per column
```

```python
# Drop ALL rows that have ANY NaN
df.dropna()
```

```python
# Drop rows where a SPECIFIC column has NaN
df.dropna(subset=["Marks"])
```

```python
# Drop rows where MULTIPLE specific columns have NaN
df.dropna(subset=["Marks", "Grade"])
```

```python
df.dropna(how="all")   # Row must be completely empty to drop
```

```python
df.dropna(axis=1)       # axis=1 means columns
```

```python
df.dropna(thresh=3)    # Keep rows that have at least 3 non-NaN values
```

```python
# By default dropna returns a NEW df, original is unchanged
df.dropna(inplace=True)    # This modifies df directly
```

```python
df.fillna(0)
```

```python
df["Marks"].fillna(0)           # Fill missing marks with 0
df["City"].fillna("Unknown")    # Fill missing city with "Unknown"
```

```python
# Fill missing marks with average marks
mean_marks = df["Marks"].mean()
df["Marks"].fillna(mean_marks)
```

```python
df["Marks"].fillna(df["Marks"].median())   # Fill with median
df["Grade"].fillna(df["Grade"].mode()[0])  # Fill with most common value
```

```python
df.fillna({
    "Marks": df["Marks"].mean(),
    "City":  "Unknown",
    "Grade": "N/A"
})
```

```python
df["City"].fillna(method="ffill")   # Copy value from row ABOVE
```

```python
df["City"].fillna(method="bfill")   # Copy value from row BELOW
```
